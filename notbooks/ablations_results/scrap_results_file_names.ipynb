```python
import os
import re
import glob
import pandas as pd

# Directory paths
job_id = '67903336'
terminal_runs_dir = f'/home/liranc6/W25/adversarial-attacks-on-deep-learning/project/terminal_runs/{job_id}'

# Build the experiment command mapping
exp_commands = []
for exp_name, values in exps.items():
    for value in values:
        exp_commands.append((exp_name, value))

# Repeat the pattern 4 times (indices 0-107)
exp_commands = exp_commands * 4

# Function to extract result file path from output file
def extract_result_path(output_file):
    """Extract the result file path from a terminal output file"""
    try:
        with open(output_file, 'r') as f:
            content = f.read()
            # Look for the JSON file path
            match = re.search(r'(comprehensive_results/subset_size-1000/[^\s]+/all_experiment_results\.json)', content)
            if match:
                return match.group(1)
    except Exception as e:
        print(f"Error reading {output_file}: {e}")
    return None

# Scrape all output files
output_files = sorted(glob.glob(f"{terminal_runs_dir}/out_job_*.txt"))

# Build the DataFrame
rows = []
for idx, output_file in enumerate(output_files):
    if idx >= len(exp_commands):
        break
    
    exp_name, exp_value = exp_commands[idx]
    result_path = extract_result_path(output_file)
    
    if result_path:
        rows.append({
            'parameter_name': exp_name,
            'val': exp_value,
            'job_idx': idx,
            'path_to_all_results': result_path
        })

# Create DataFrame
results_df = pd.DataFrame(rows)

# Set pandas to display full column width
pd.set_option('display.max_colwidth', None)

# Extract timestamp from path
results_df['timestamp'] = results_df['path_to_all_results'].apply(lambda x: x.split('/')[2])
# results_df['dir_path'] = results_df['path_to_all_results'].apply(lambda x: '/'.join(x.split('/')[:-1]))

print(f"Found {len(results_df)} results")
results_df
```
